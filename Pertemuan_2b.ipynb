```python
#Import Library & Deklarasi Path
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Path disesuaikan dengan struktur folder Google Drive kamu
path_gambar = '/content/drive/MyDrive/PCVK/KTM dewi.jpeg'
```

```python
# Read image menggunakan OpenCV (Default BGR)
img = cv2.imread(path_gambar)

# Konversi BGR ke RGB agar warna sesuai di Matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Tampilkan dengan Plotting Matplotlib
plt.figure(figsize=(6, 4))
plt.imshow(img_rgb)
plt.title("Citra Format RGB")
plt.axis('on')
plt.show()
```

```python
img_gray = cv2.imread(path_gambar, cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(10, 5))

# Tanpa cmap
plt.subplot(1, 2, 1)
plt.imshow(img_gray)
plt.title("Tanpa cmap")

# Dengan cmap 'gray'
plt.subplot(1, 2, 2)
plt.imshow(img_gray, cmap="gray")
plt.title("Dengan cmap")

plt.show()
```

```python
# Resize ke ukuran 400x250
img_small = cv2.resize(img_rgb, (400, 250))

# Flip Horisontal (1)
img_flip = cv2.flip(img_small, 1)

# Tampilkan dengan Canvas Lebar
fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111)
ax.imshow(img_flip)
plt.title("Gambar Hasil Flip Horisontal")
plt.show()
```

```python
# 1. Buat Canvas Hitam (512x512 piksel, 3 channel)
black_img = np.zeros(shape=(512, 512, 3), dtype=np.int16)

# 2. Menambahkan persegi panjang hijau
cv2.rectangle(black_img, pt1=(384, 8), pt2=(510, 150), color=(0, 255, 0), thickness=10)
plt.figure(figsize=(4, 4))
plt.imshow(black_img)
plt.title("1. Rectangle Hijau")
plt.show()

# 3. Menambahkan persegi biru
cv2.rectangle(black_img, pt1=(200, 200), pt2=(300, 300), color=(0, 0, 255), thickness=15)
plt.figure(figsize=(4, 4))
plt.imshow(black_img)
plt.title("2. Rectangle Biru")
plt.show()

# 4. Menambahkan lingkaran merah
cv2.circle(black_img, center=(100, 100), radius=50, color=(255, 0, 0), thickness=8)
plt.figure(figsize=(4, 4))
plt.imshow(black_img)
plt.title("3. Circle Merah")
plt.show()

# 5. Menambahkan garis diagonal putih
cv2.line(black_img, pt1=(0, 0), pt2=(512, 512), color=(255, 255, 255), thickness=5)
plt.figure(figsize=(4, 4))
plt.imshow(black_img)
plt.title("4. Line Diagonal")
plt.show()

# 6. Menambahkan Teks 'Hello'
font = cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(black_img, text='Hello', org=(10, 500), fontFace=font, fontScale=4, color=(255, 255, 0), thickness=2, lineType=cv2.LINE_AA)
plt.figure(figsize=(4, 4))
plt.imshow(black_img)
plt.title("5. Teks Hello")
plt.show()
```

== PERTANYAAN PRAKTIKUM D3 ==
1. Perbedaan Tampilan Tanpa vs Dengan Matplotlib
- Tanpa Matplotlib (cv2.imshow / cv2_imshow): Menampilkan citra murni tanpa sumbu koordinat (axis).
- Dengan Matplotlib (plt.imshow): Menampilkan citra bersumbu koordinat X dan Y untuk melihat titik piksel serta mendukung pemakaian colormap (cmap)

2. Perbedaan & Pengaruh Black Image int16 vs int32
- Perbedaan: int16 berukuran 16-bit (2 byte/elemen), sedangkan int32 berukuran 32-bit (4 byte/elemen).
- Pengaruh: Secara visual keduanya tampak persis sama. Namun, int32 memakai memori 2x lebih besar dan wajib digunakan pada fungsi OpenCV tertentu (seperti cv2.polylines) untuk mencegah type error atau overflow.

3. Kegunaan google.colab.patches.cv2_imshow
Menggantikan cv2.imshow() bawaan OpenCV yang menyebabkan crash kernel di Google Colab, sehingga citra bisa tampil langsung di notebook.

4. Kegunaan skimage.io
Menangani fungsi Input/Output citra, seperti mengunduh dan membaca citra secara langsung dari URL web ke bentuk array.

```python
# Cell 1: Import Library & Load Gambar
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Sesuaikan dengan path file gambar kamu
path_gambar = '/content/drive/MyDrive/PCVK/KTM dewi.jpeg'

# Read image & konversi ke RGB
img = cv2.imread(path_gambar)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
# Tugas 1 - Mengubah Ukuran Canvas (Figsize)
# Penjelasan: Mengubah figsize HANYA memperbesar tampilan visual plot di Colab,
# TIDAK mengubah jumlah/resolusi piksel asli dari citra.
plt.figure(figsize=(12, 8))
plt.imshow(img_rgb)
plt.title("Tugas 1: Pengaruh Figsize pada Tampilan Visual")
plt.show()
```

```python
# Tugas 2 - Menampilkan Channel Red-Blue & Green-Blue
# Channel Red-Blue (Matikan channel Green/index 1)
img_RB = img_rgb.copy()
img_RB[:, :, 1] = 0

# Channel Green-Blue (Matikan channel Red/index 0)
img_GB = img_rgb.copy()
img_GB[:, :, 0] = 0

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img_RB)
plt.title("Channel Red-Blue")

plt.subplot(1, 2, 2)
plt.imshow(img_GB)
plt.title("Channel Green-Blue")

plt.show()
```

```python
# Tugas 3 - Slicing Citra Baris 20-115 & Kolom 25-120
crop_tugas3 = img_rgb[20:115, 25:120]

plt.figure(figsize=(4, 4))
plt.imshow(crop_tugas3)
plt.title("Tugas 3: Slicing Spasial Baris 20-115 & Kolom 25-120")
plt.show()
```

```python
# Tugas 4 - Slicing Baris 5-30, Semua Kolom, Channel Red Saja
crop_tugas4 = img_rgb[5:30, :, 0]  # Channel 0 = Red

plt.figure(figsize=(8, 2))
plt.imshow(crop_tugas4, cmap='gray')
plt.title("Tugas 4: Channel Red Saja (Grayscale)")
plt.show()
```

```python
# Tugas 5 - Membuat 5 Kotak Ukuran & Warna Acak (Random)
img_tugas5 = img_rgb.copy()
h, w, _ = img_tugas5.shape

# Generate 5 kotak secara acak
for _ in range(5):
    pt1 = (np.random.randint(0, w - 50), np.random.randint(0, h - 50))
    pt2 = (pt1[0] + np.random.randint(20, 80), pt1[1] + np.random.randint(20, 80))
    color = (np.random.randint(0, 256), np.random.randint(0, 256), np.random.randint(0, 256))
    cv2.rectangle(img_tugas5, pt1, pt2, color, -1)

plt.figure(figsize=(6, 4))
plt.imshow(img_tugas5)
plt.title("Tugas 5: 5 Kotak dengan Ukuran & Warna Acak")
plt.show()
```

```python
# Tugas 6 - Menampilkan Citra Terbalik (Flip Vertikal & Horisontal)
# Flip -1 melakukan pemutaran vertikal sekaligus horisontal
img_tugas6 = cv2.flip(img_rgb, -1)

plt.figure(figsize=(6, 4))
plt.imshow(img_tugas6)
plt.title("Tugas 6: Gambar Terbalik")
plt.show()
```

```python
# Tugas 7 - Menambahkan Rectangle & Circle di Area Wajah
img_tugas7 = img_rgb.copy()

# Koordinat penyesuaian area foto wajah pada KTM
cv2.rectangle(img_tugas7, (50, 450), (350, 850), (255, 0, 0), 5)
cv2.circle(img_tugas7, (200, 650), 180, (0, 255, 0), 4)

plt.figure(figsize=(6, 4))
plt.imshow(img_tugas7)
plt.title("Tugas 7: Rectangle & Circle pada Area Wajah")
plt.show()
```

```python
# Tugas 8 - Rectangle di Sudut Bawah Kiri pada Channel B (Blue)
# Ambil Channel B saja (Index 2)
img_B = img_rgb[:, :, 2].copy()
h_b, w_b = img_B.shape

# Gambar persegi panjang di sudut bawah kiri
cv2.rectangle(img_B, (20, h_b - 150), (250, h_b - 20), (255, 255, 255), 4)

plt.figure(figsize=(6, 4))
plt.imshow(img_B, cmap='gray')
plt.title("Tugas 8: Channel B dengan Rectangle Sudut Bawah Kiri")
plt.show()
```

```python
# Tugas 9 - Menambahkan Tulisan Nama File Citra
# Konversi ke 3 channel agar bisa diberi teks berwarna
img_tugas9 = cv2.cvtColor(img_B, cv2.COLOR_GRAY2RGB)

cv2.putText(img_tugas9, "File: KTM dewi.jpeg", (30, h_b - 50),
            cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 0), 2, cv2.LINE_AA)

plt.figure(figsize=(6, 4))
plt.imshow(img_tugas9)
plt.title("Tugas 9: Ditambah Tulisan Nama File Citra")
plt.show()
```
